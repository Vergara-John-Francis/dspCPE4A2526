```python
import cv2
import numpy as np

# Define the value of Philippine coins (for example)
coin_values = {
    'PHP 1': 1,
    'PHP 5': 5,
    'PHP 10': 10,
    'PHP 0.25': 0.25,  # 25 cents
    'PHP 0.01': 0.01   # 1 cent
}

# Function to detect coins and classify them by size
def detect_coins(frame):
    # Convert to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    # Apply Gaussian blur to reduce noise and improve circle detection
    blurred = cv2.GaussianBlur(gray, (15, 15), 0)
    
    # Use HoughCircles to detect circles (coins)
    circles = cv2.HoughCircles(blurred, cv2.HOUGH_GRADIENT, dp=1.2, minDist=50, param1=50, param2=30, minRadius=15, maxRadius=40)
    
    detected_coins = []
    total_value = 0
    
    if circles is not None:
        circles = np.round(circles[0, :]).astype("int")
        
        for (x, y, r) in circles:
            # Draw the circle in the output image
            cv2.circle(frame, (x, y), r, (0, 255, 0), 4)
            cv2.rectangle(frame, (x - 5, y - 5), (x + 5, y + 5), (0, 128, 255), -1)
            
            # Classify the coin based on radius
            if r >= 15 and r < 20:
                detected_coins.append('PHP 1')  # PHP 1 coin
                total_value += coin_values['PHP 1']
            elif r >= 20 and r < 25:
                detected_coins.append('PHP 5')  # PHP 5 coin
                total_value += coin_values['PHP 5']
            elif r >= 25:
                detected_coins.append('PHP 10')  # PHP 10 coin
                total_value += coin_values['PHP 10']
            else:
                detected_coins.append('PHP 0.25')  # PHP 0.25 coin
                total_value += coin_values['PHP 0.25']

    return frame, detected_coins, total_value

# Main loop for real-time processing
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()  # Read a frame from the video capture
    if not ret:
        break
    
    # Detect coins and get updated frame with detection
    result_frame, detected_coins, total_value = detect_coins(frame)

    # Display the number of detected coins and total value
    cv2.putText(result_frame, f"Detected Coins: {len(detected_coins)}", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.putText(result_frame, f"Total Value: PHP {total_value:.2f}", (10, 60), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

    # Show the resulting frame
    cv2.imshow("Coin Detection", result_frame)

    # Exit if the 'q' key is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources and close the window
cap.release()
cv2.destroyAllWindows()
```
