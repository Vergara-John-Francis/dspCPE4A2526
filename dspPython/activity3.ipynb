```python
import cv2
import numpy as np

coin_values = {'PHP 1': 1, 'PHP 5': 5, 'PHP 10': 10, 'PHP 0.25': 0.25, 'PHP 0.01': 0.01}

def detect_coins(frame):
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (15, 15), 0)
    circles = cv2.HoughCircles(blurred, cv2.HOUGH_GRADIENT, dp=1.2, minDist=50, param1=50, param2=30, minRadius=15, maxRadius=40)
    detected_coins = []
    total_value = 0
    if circles is not None:
        circles = np.round(circles[0, :]).astype("int")
        for (x, y, r) in circles:
            cv2.circle(frame, (x, y), r, (0, 255, 0), 4)
            cv2.rectangle(frame, (x - 5, y - 5), (x + 5, y + 5), (0, 128, 255), -1)
            if r >= 15 and r < 20:
                detected_coins.append('PHP 1')
                total_value += coin_values['PHP 1']
            elif r >= 20 and r < 25:
                detected_coins.append('PHP 5')
                total_value += coin_values['PHP 5']
            elif r >= 25:
                detected_coins.append('PHP 10')
                total_value += coin_values['PHP 10']
            else:
                detected_coins.append('PHP 0.25')
                total_value += coin_values['PHP 0.25']
    return frame, detected_coins, total_value

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    result_frame, detected_coins, total_value = detect_coins(frame)
    cv2.putText(result_frame, f"Detected Coins: {len(detected_coins)}", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.putText(result_frame, f"Total Value: PHP {total_value:.2f}", (10, 60), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
    cv2.imshow("Coin Detection", result_frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
